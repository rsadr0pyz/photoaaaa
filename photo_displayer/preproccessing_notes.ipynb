```python
import json
import numpy as np
import pandas as pd
import os
import math
from PIL import Image
```

```python
with open("../config.json") as json_file:
    CONFIG_FILE = json.load(json_file)

PAD_25_PATH = os.path.join(CONFIG_FILE["pad-25-dataset-path"], "anonymous-metadata.csv")  
REDDIT_PATH = os.path.join(CONFIG_FILE["reddit-dataset-path"], "images") 
```

Utility

```python
# Makes a concise table for display
def make_table(values, fill, cols = 10):
    rows = math.ceil(len(values) / cols)
    fill_count = rows * cols - len(values)
    final = np.append(values, [fill] * fill_count)
    final = final.reshape([rows, cols])
    return pd.DataFrame(final)


cluster_targets = {
    "C43": "P1",
    "D03": "P1",
    "D22": "P1",

    "C80": "P2",
    "C44": "P2",
    "D04": "P2",
    "L75": "P2",  
    "D23": "P2",    

    "L57": "P3",
    "L25": "P3",
    "L30": "P3",
    "L98": "P3",    

    "L78": "P4",
    "L82": "P4",

    "L70": "P5",
    "00": "P5"
}
```

# PAD 25 Dataset

```python
# Load relevant data

dtype = {
    "img-id": str,
    "macroCIDDiagnostic": "category",
    "img-src": "category"
}
relevant_columns = dtype.keys()

p25_dataset = pd.read_csv(PAD_25_PATH, dtype=dtype, usecols=relevant_columns)

# Filter Dematoscopic images
clinical_source_mask = (p25_dataset["img-src"] == "CLINICAL")
p25_dataset = p25_dataset[clinical_source_mask]
p25_dataset = p25_dataset.drop(columns=["img-src"])

# Add file extensions
p25_dataset["img-id"] += ".png"

p25_dataset
```

```python
#Clustering

p25_dataset["macroCIDDiagnostic"] = p25_dataset["macroCIDDiagnostic"].map(cluster_targets)
p25_dataset  = p25_dataset.dropna()
```

```python
unique_values = p25_dataset["macroCIDDiagnostic"].unique()
fill = "-"
assert not (fill in unique_values)

make_table(unique_values, fill)
```

# REDDIT Dataset

```python
# Load relevant data

file_list = []
for file in os.listdir(REDDIT_PATH):
    file_list.append(file)

reddit_dataset = pd.DataFrame({'img-id': file_list, 'macroCIDDiagnostic': "P5"})

reddit_dataset
```

```python
# Clustering

reddit_dataset["macroCIDDiagnostic"] = reddit_dataset["macroCIDDiagnostic"].map(cluster_targets)
reddit_dataset = reddit_dataset.dropna()
```

```python
unique_values = reddit_dataset["macroCIDDiagnostic"].unique()
fill = "-"
assert not (fill in unique_values)

make_table(unique_values, fill, cols=5)
```
